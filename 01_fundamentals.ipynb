basic variables & data types

```python
customer_id = 1001
customer_name = "Ras"
monthly_income = 55_000.50
is_active = True
prediction = None

print("Customer ID:", customer_id)
print("Customer name:", customer_name)
print("Monthly income:", monthly_income)
print("Active:", is_active)
print("Prediction:", prediction)

print("\nData types")
print(type(customer_id))
print(type(customer_name))
print(type(monthly_income))
print(type(is_active))
print(type(prediction))
```

type conversion

```python
age_text = "28"
income_text = "55000.50"
purchased_text = "yes"

age = int(age_text)
income = float(income_text)
purchased = purchased_text.strip() == "yes"

print(age, type(age))
print(income, type(income))
print(purchased, type(purchased))
```

Arithmetic operators

```python
first_value = 100
second_value = 30

print("Addition:",first_value + second_value)
print("Subtraction:", first_value - second_value)
print("Multiplication:", first_value * second_value)
print("Division:", first_value / second_value)
print("Floor division:", first_value // second_value)
print("Remainder:", first_value % second_value)
print("Power:", first_value ** 2)
```

Comparison and logical operators

```python
accuracy = 0.89
precision = 0.86
recall = 0.83

print("Accuracy acceptable:", accuracy >= 0.85)
print("Precision equals 0.86:", precision == 0.86)
print("Recall not equal to 0.80:", recall != 0.80)

deployment_ready = (
    accuracy >= 0.85
    and precision >= 0.80
    and recall >= 0.80
)

print("Deployment ready:", deployment_ready)
```

== versus is

```python
list_one = [1, 2, 3]
list_two = [1, 2, 3]
list_three = list_one

print("Same values:", list_one == list_two)
print("Same object:", list_one is list_two)
print("Same object reference:", list_one is list_three)

result = None

if result is None:
    print("Result is unavailable")
```

String cleaning

```python
review = "   Excellent PRODUCT!!!   "

cleaned_review = (
    review
    .strip()
    .lower()
    .replace("!", "")
)

print("Original:", repr(review))
print("Cleaned:", cleaned_review)
```

Common string operations

```python
text = "machine learning requires hands-on practice"

print("Upper:", text.upper())
print("Lower:", text.lower())
print("Title:", text.title())
print("Starts with machine:", text.startswith("machine"))
print("Ends with practice:", text.endswith("practice"))
print("Position of learning:", text.find("learning"))
print("Replacement:", text.replace("practice", "projects"))
```

Split and join

```python
skills_text = "python,numpy,pandas,scikit-learn,pytorch"

skills = skills_text.split(",")

print("Skills list:", skills)
print("Joined skills:", " | ".join(skills))
```
