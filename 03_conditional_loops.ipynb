Conditions

```python
accuracy = 0.89
recall = 0.81

if accuracy >= 0.90 and recall >= 0.85:
    status = "Ready for deployment"
elif accuracy >= 0.85:
    status = "Requires further validation"
else:
    status = "Retrain the model"

print(status)
```

Credit-risk classification

```python
credit_score = 710

if credit_score is None:
    risk = "Unable to assess"
elif credit_score >= 750:
    risk = "Low risk"
elif credit_score >= 650:
    risk = "Medium risk"
else:
    risk = "High risk"

print("Risk category:", risk)
```

For loop

```python
model_scores = [0.81, 0.86, 0.91, 0.78]

for i in model_scores:
    if i >= 0.85:
        print(i, "Accepted")
    else:
        print(i, "Rejected")
```

Enumerate

```python
models = [
    "Logistic Regression",
    "Random Forest",
    "XGBoost"
]

for position, model in enumerate(models, start=1):
    print(f"{position}. {model}")
```

Zip

```python
models = [
    "Logistic Regression",
    "Random Forest",
    "XGBoost"
]

scores = [0.84, 0.89, 0.92]

for model, score in zip(models, scores):
    print(f"{model}: {score:.2%}")
```

Break and continue

```python
values = [10, None, 20, -999, 30]

for value in values:
    if value is None:
        print("Skipping missing value")
        continue

    if value == -999:
        print("Sentinel detected. Stopping.")
        break

    print("Processing:", value)
```

While loop

```python
epoch = 1
maximum_epochs = 5

while epoch <= maximum_epochs:
    print(f"Training epoch {epoch}")
    epoch += 1
```

Manual accuracy

```python
actual = [1, 0, 1, 1, 0, 0, 1]
predicted = [1, 0, 0, 1, 1, 0, 1]

correct_predictions = 0

for actual_value, predicted_value in zip(actual, predicted):
    if actual_value == predicted_value:
        correct_predictions += 1

accuracy = correct_predictions / len(actual)

print("Correct:", correct_predictions)
print("Total:", len(actual))
print(f"Accuracy: {accuracy:.2%}")
```
