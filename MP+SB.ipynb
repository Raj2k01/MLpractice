import libraries

import pandas as pd

```python
#Sample Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Age":[21,22,23,24,25,26],
    "Salary":[30000,35000,40000,42000,45000,50000],
    "Experience":[1,2,3,4,5,6],
    "Department":["IT","IT","HR","HR","Finance","Finance"]
})
```

MATPLOTLIB

1. Line Plot - Used for trends

```python
plt.plot(df["Age"], df["Salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()

print(df)
```

2. Scatter Plot - Most important ML visualization

Use Cases:

Relationship between variables
Linear Regression analysis

```python
plt.scatter(
    df["Experience"],
    df["Salary"]
)

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
```

3. Bar Chart

```python
plt.bar(
    df["Department"],
    [2,2,2]
)

plt.title("Department Count")

plt.show() #error len mismatch error
```

4. Histogram - Very Important, Shows distribution

Interview:

How do you check if data is normally distributed?

Histogram.

```python
plt.hist(
    df["Salary"],
    bins=5
)

plt.title("Salary Distribution")

plt.show()
print("\n", df)
```

5. Box Plot - Extremely Important, Use: Outlier Detection

```python
plt.boxplot(df["Salary"])

plt.title("Salary Boxplot")

plt.show()
```

6. Multiple Lines

```python
plt.plot(df["Age"], df["Salary"])

plt.plot(df["Age"], df["Experience"])

plt.legend([
    "Salary",
    "Experience"
])

plt.show()
```

7. Figure Size

```python
plt.figure(figsize=(8,5))

plt.plot(df["Age"],df["Salary"])

plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

8. Save Plot - ML projects frequently save visualizations

```python
plt.plot(df["Age"],df["Salary"])

plt.xlabel("Age")
plt.ylabel("Salary")

plt.savefig("salary_plot.png")
```

SEABORN - More beautiful and ML-oriented

1. Scatter Plot

```python
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary"
)

plt.show()
```

2. Line Plot

```python
sns.lineplot(
    data=df,
    x="Age",
    y="Salary"
)

plt.show()
```

3. Histogram - Very common during EDA

```python
sns.histplot(
    df["Salary"],
    kde=True
)

plt.show()
```

4. Box Plot

Most important, Outlier detection

```python
sns.boxplot(
    y=df["Salary"]
)

plt.show()
```

5. Count Plot

Very important

```python
sns.countplot(
    data=df,
    x="Department"
)

plt.show()
```

6. Violin Plot - Interview Favorite

Shows:

Distribution
Density
Outliers

```python
sns.violinplot(
    y=df["Salary"]
)

plt.show()
```

7. Pair Plot - ML Engineer Favorite

One command.

Very important for EDA.

```python
sns.pairplot(df)

plt.show()
```

8. Correlation Heatmap - MOST IMPORTANT SEABORN PLOT

Use Cases:

Feature Selection
Detect multicollinearity
Understand relationships

Interview favorite.

```python
corr = df.corr(
    numeric_only=True
)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

9. Regression Plot - Important for ML

Shows:

Scatter plot
Regression line

Useful for Linear Regression

```python
sns.regplot(
    data=df,
    x="Experience",
    y="Salary"
)

plt.show()
```

10. KDE Plot - Kernel Density Estimate

Alternative to histogram

```python
sns.kdeplot(
    df["Salary"],
    fill=True
)

plt.show()
```

ML Mini Project 1: Employee Salary Analysis

Goal:

Distribution

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("employee_data.csv")

sns.histplot(
    df["Salary"],
    kde=True
)

plt.show()

sns.boxplot(
    y=df["Salary"]
)

plt.show()
```

ML Mini Project 2: Feature Relationship

Goal:

Check correlation before modeling

```python
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary"
)

plt.show()

sns.regplot(
    data=df,
    x="Experience",
    y="Salary"
)

plt.show()
```

ML Mini Project 3: Correlation Heatmap

Goal:

Feature selection
Remove redundant columns

```python
corr = df.corr(
    numeric_only=True
)

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```
