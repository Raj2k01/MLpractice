ML Mini Project 1: Student Analysis

```python
import pandas as pd

students = pd.DataFrame({
    "Name":["A","B","C","D"],
    "Math":[90,80,70,95],
    "Science":[85,75,80,90]
})

students["Average"] = (
    students["Math"]
    +
    students["Science"]
)/2

print(students)

print("\n",
    students.sort_values(
        "Average",
        ascending=False
    )
)
```

ML Mini Project 2: Missing Value Handling

```python
import pandas as pd
import numpy as np

employees = pd.DataFrame({
    "Age":[25,30,np.nan,40],
    "Salary":[30000,40000,50000,np.nan]
})

print(employees.isnull().sum())

employees["Age"] = employees["Age"].fillna(
    employees["Age"].mean()
)

employees["Salary"] = employees["Salary"].fillna(
    employees["Salary"].mean()
)

print("\n",employees)
```
