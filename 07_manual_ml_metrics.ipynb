Metrics function

```python
def classification_metrics(actual, predicted):
    if not isinstance(actual, list):
        raise TypeError("Actual must be a list.")

    if not isinstance(predicted, list):
        raise TypeError("Predicted must be a list.")

    if len(actual) != len(predicted):
        raise ValueError(
            "Actual and predicted must have equal length."
        )

    if not actual:
        raise ValueError("Input lists cannot be empty.")

    valid_labels = {0, 1}

    if not set(actual).issubset(valid_labels):
        raise ValueError(
            "Actual values must contain only 0 and 1."
        )

    if not set(predicted).issubset(valid_labels):
        raise ValueError(
            "Predicted values must contain only 0 and 1."
        )

    true_positive = 0
    true_negative = 0
    false_positive = 0
    false_negative = 0

    for actual_value, predicted_value in zip(
        actual,
        predicted
    ):
        if actual_value == 1 and predicted_value == 1:
            true_positive += 1

        elif actual_value == 0 and predicted_value == 0:
            true_negative += 1

        elif actual_value == 0 and predicted_value == 1:
            false_positive += 1

        elif actual_value == 1 and predicted_value == 0:
            false_negative += 1

    total = len(actual)

    accuracy = (
        true_positive + true_negative
    ) / total

    precision_denominator = (
        true_positive + false_positive
    )

    precision = (
        true_positive / precision_denominator
        if precision_denominator > 0
        else 0
    )

    recall_denominator = (
        true_positive + false_negative
    )

    recall = (
        true_positive / recall_denominator
        if recall_denominator > 0
        else 0
    )

    specificity_denominator = (
        true_negative + false_positive
    )

    specificity = (
        true_negative / specificity_denominator
        if specificity_denominator > 0
        else 0
    )

    f1_denominator = precision + recall

    f1_score = (
        2 * precision * recall / f1_denominator
        if f1_denominator > 0
        else 0
    )

    balanced_accuracy = (
        recall + specificity
    ) / 2

    return {
        "true_positive": true_positive,
        "true_negative": true_negative,
        "false_positive": false_positive,
        "false_negative": false_negative,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "specificity": specificity,
        "f1_score": f1_score,
        "balanced_accuracy": balanced_accuracy
    }
```

Run the function

```python
actual = [
    1, 0, 1, 1, 0,
    1, 0, 0, 1, 0
]

predicted = [
    1, 0, 0, 1, 1,
    1, 0, 0, 1, 1
]

metrics = classification_metrics(
    actual,
    predicted
)

for metric_name, metric_value in metrics.items():
    if isinstance(metric_value, float):
        print(f"{metric_name}: {metric_value:.4f}")
    else:
        print(f"{metric_name}: {metric_value}")
```

Test the function

```python
perfect_metrics = classification_metrics(
    [1, 0, 1, 0],
    [1, 0, 1, 0]
)

assert perfect_metrics["true_positive"] == 2
assert perfect_metrics["true_negative"] == 2
assert perfect_metrics["false_positive"] == 0
assert perfect_metrics["false_negative"] == 0
assert perfect_metrics["accuracy"] == 1.0
assert perfect_metrics["precision"] == 1.0
assert perfect_metrics["recall"] == 1.0
assert perfect_metrics["specificity"] == 1.0
assert perfect_metrics["f1_score"] == 1.0
assert perfect_metrics["balanced_accuracy"] == 1.0

print("All metric tests passed")
```
