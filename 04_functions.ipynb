Basic cleaning function

```python
def clean_category(value):
    if value is None:
        return "unknown"

    cleaned_value = str(value).strip().lower()

    if cleaned_value == "":
        return "unknown"

    return cleaned_value


print(clean_category(" PREMIUM "))
print(clean_category(""))
print(clean_category(None))
```

Function with type hints

```python
def calculate_mean(values: list[float]) -> float:
    """Calculate the mean of a non-empty numeric list."""

    if not values:
        raise ValueError("Values cannot be empty.")

    return sum(values) / len(values)


scores = [0.82, 0.88, 0.91]

print(calculate_mean(scores))
```

Default and keyword arguments

```python
def classify_probability(
    probability: float,
    threshold: float = 0.50
) -> int:
    return 1 if probability >= threshold else 0


print(classify_probability(0.65))
print(classify_probability(0.65, 0.70))

print(
    classify_probability(
        probability=0.85,
        threshold=0.80
    )
)
```

Return multiple values

```python
def summarize(values):
    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)

    return minimum, maximum, average


lowest, highest, mean = summarize([10, 20, 30, 40])

print("Lowest:", lowest)
print("Highest:", highest)
print("Mean:", mean)
```

* args

```python
def calculate_average(*values):
    if not values:
        return 0

    return sum(values) / len(values)


print(calculate_average(10, 20, 30))
print(calculate_average(10, 20, 30, 40, 50))
```

** args

```python
def create_model_config(**parameters):
    return parameters


config = create_model_config(
    model="XGBoost",
    max_depth=5,
    learning_rate=0.1,
    n_estimators=100
)

print(config)
```

Lambda and sorted

```python
experiments = [
    {"model": "Logistic Regression", "score": 0.84},
    {"model": "Random Forest", "score": 0.90},
    {"model": "XGBoost", "score": 0.92}
]

ranked_experiments = sorted(
    experiments,
    key=lambda experiment: experiment["score"],
    reverse=True
)

for experiment in ranked_experiments:
    print(experiment)
```

Map and filter

```python
raw_values = ["10", "20", "30", "40"]

numeric_values = list(map(int, raw_values))

scores = [0.72, 0.91, 0.84, 0.95]

high_scores = list(
    filter(lambda score: score >= 0.85, scores)
)

print("Numeric values:", numeric_values)
print("High scores:", high_scores)
```
