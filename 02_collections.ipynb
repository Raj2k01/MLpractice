Lists

```python
model_scores = [0.82, 0.91, 0.87, 0.89]

model_scores.append(0.93)
model_scores.insert(1, 0.85)

print("After adding:", model_scores)

model_scores.remove(0.85)
removed_score = model_scores.pop()

print("Popped score:", removed_score)
print("Remaining scores:", model_scores)

model_scores.sort(reverse=True)

print("Sorted scores:", model_scores)
print("Best:", max(model_scores))
print("Worst:", min(model_scores))
print("Average:", sum(model_scores) / len(model_scores))
```

List slicing

```python
values = [10, 20, 30, 40, 50, 60]

print("First:", values[0])
print("Last:", values[-1])
print("Index 1 to 3:", values[1:4])
print("First three:", values[:3])
print("Every second value:", values[::2])
print("Reversed:", values[::-1])
```

Analyze predictions

```python
predictions = [1, 0, 1, 1, 0, 1, 0, 1]

total = len(predictions)
positive = predictions.count(1)
negative = predictions.count(0)
positive_rate = positive / total

print("Total:", total)
print("Positive:", positive)
print("Negative:", negative)
print(f"Positive rate: {positive_rate:.2%}")
```

Feature validation using sets

```python
training_features = {
    "age",
    "income",
    "credit_score",
    "city"
}

inference_features = {
    "age",
    "income",
    "location"
}

missing_features = training_features - inference_features
unexpected_features = inference_features - training_features
common_features = training_features & inference_features

print("Missing:", missing_features)
print("Unexpected:", unexpected_features)
print("Common:", common_features)
```

Dictionaries

```python
model_metrics = {
    "model": "Random Forest",
    "accuracy": 0.91,
    "precision": 0.89,
    "recall": 0.87
}

model_metrics["f1_score"] = 0.88
model_metrics["version"] = 1

print("Accuracy:", model_metrics["accuracy"])
print("F1:", model_metrics.get("f1_score"))

for key, value in model_metrics.items():
    print(f"{key}: {value}")
```

Nested dictionary

```python
experiment = {
    "model": {
        "name": "XGBoost",
        "version": 1
    },
    "parameters": {
        "max_depth": 5,
        "learning_rate": 0.1,
        "n_estimators": 100
    },
    "metrics": {
        "accuracy": 0.92,
        "precision": 0.90,
        "recall": 0.88,
        "f1_score": 0.89
    }
}

print("Model:", experiment["model"]["name"])
print("Depth:", experiment["parameters"]["max_depth"])
print("F1:", experiment["metrics"]["f1_score"])
```

Comprehensions

```python
numbers = [1, 2, 3, 4, 5, 6]

squares = [number ** 2 for number in numbers]

even_numbers = [
    number
    for number in numbers
    if number % 2 == 0
]

print("Squares:", squares)
print("Even numbers:", even_numbers)
```

ML-style comprehensions

```python
scores = [0.72, 0.91, 0.84, 0.95, 0.68]

accepted_scores = [
    score
    for score in scores
    if score >= 0.85
]

features = ["age", "income", "credit_score"]

feature_positions = {
    feature: index
    for index, feature in enumerate(features)
}

print("Accepted scores:", accepted_scores)
print("Feature positions:", feature_positions)
```
