Data-cleaning class

```python
class DataCleaner:
    def __init__(
        self,
        missing_numeric_value=0,
        missing_category="unknown"
    ):
        self.missing_numeric_value = missing_numeric_value
        self.missing_category = missing_category

    def clean_numeric(self, value):
        if value is None or value == "":
            return self.missing_numeric_value

        try:
            return float(value)
        except (TypeError, ValueError):
            return self.missing_numeric_value

    def clean_category(self, value):
        if value is None:
            return self.missing_category

        cleaned_value = str(value).strip().lower()

        if cleaned_value == "":
            return self.missing_category

        return cleaned_value

    def __str__(self):
        return (
            f"DataCleaner("
            f"numeric={self.missing_numeric_value}, "
            f"category='{self.missing_category}')"
        )


cleaner = DataCleaner()

print(cleaner)
print(cleaner.clean_numeric("52000"))
print(cleaner.clean_numeric("invalid"))
print(cleaner.clean_category(" PREMIUM "))
```

ML-style class

```python
class ThresholdClassifier:
    def __init__(self, threshold=0.50):
        self.threshold = threshold
        self.is_trained = False

    def fit(self, training_data):
        if not training_data:
            raise ValueError("Training data cannot be empty.")

        self.is_trained = True
        print(f"Trained using {len(training_data)} records")

        return self

    def predict(self, probabilities):
        if not self.is_trained:
            raise RuntimeError("Train before prediction.")

        return [
            1 if probability >= self.threshold else 0
            for probability in probabilities
        ]


model = ThresholdClassifier(threshold=0.60)

model.fit([1, 2, 3, 4])

predictions = model.predict(
    [0.30, 0.65, 0.90, 0.55]
)

print(predictions)
```

Generator for data batches

```python
def batch_generator(data, batch_size):
    if batch_size <= 0:
        raise ValueError("Batch size must be positive.")

    for start_index in range(
        0,
        len(data),
        batch_size
    ):
        end_index = start_index + batch_size

        yield data[start_index:end_index]


records = list(range(1, 11))

for batch in batch_generator(records, 3):
    print(batch)
```

Assertions

```python
def normalize_score(score):
    if not isinstance(score, (int, float)):
        raise TypeError("Score must be numeric.")

    if not 0 <= score <= 100:
        raise ValueError("Score must be between 0 and 100.")

    return score / 100


assert normalize_score(80) == 0.8
assert normalize_score(0) == 0
assert normalize_score(100) == 1

print("All tests passed")
```
