Safe numeric conversion

```python
def safe_int(value):
    try:
        return int(value)
    except (TypeError, ValueError):
        return None


def safe_float(value):
    try:
        return float(value)
    except (TypeError, ValueError):
        return None


print(safe_int("28"))
print(safe_int("invalid"))
print(safe_float("55000.50"))
print(safe_float(None))
```

Raising exceptions

```python
def validate_probability(value):
    if not isinstance(value, (int, float)):
        raise TypeError("Probability must be numeric.")

    if not 0 <= value <= 1:
        raise ValueError("Probability must be between 0 and 1.")

    return value


print(validate_probability(0.85))
```

Write and read a text file

```python
training_log = """
Model: XGBoost
Dataset: Wine Quality
Status: Training completed
Accuracy: 0.89
"""

with open(
    "training_log.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(training_log)

with open(
    "training_log.txt",
    "r",
    encoding="utf-8"
) as file:
    content = file.read()

print(content)
```

CSV writing

```python
import csv


customers = [
    {
        "customer_id": 1,
        "age": 28,
        "income": 50_000
    },
    {
        "customer_id": 2,
        "age": 35,
        "income": 72_000
    },
    {
        "customer_id": 3,
        "age": 42,
        "income": 68_000
    }
]

with open(
    "customers.csv",
    "w",
    newline="",
    encoding="utf-8"
) as file:
    fieldnames = [
        "customer_id",
        "age",
        "income"
    ]

    writer = csv.DictWriter(
        file,
        fieldnames=fieldnames
    )

    writer.writeheader()
    writer.writerows(customers)

print("CSV created")
```

CSV reading

```python
import csv


loaded_customers = []

with open(
    "customers.csv",
    "r",
    encoding="utf-8"
) as file:
    reader = csv.DictReader(file)

    for row in reader:
        cleaned_row = {
            "customer_id": int(row["customer_id"]),
            "age": int(row["age"]),
            "income": float(row["income"])
        }

        loaded_customers.append(cleaned_row)

print(loaded_customers)
```

JSON writing and reading

```python
import json


prediction = {
    "customer_id": "C001",
    "prediction": "high_risk",
    "probability": 0.87,
    "model_version": "1.0"
}

with open(
    "prediction.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        prediction,
        file,
        indent=4
    )

with open(
    "prediction.json",
    "r",
    encoding="utf-8"
) as file:
    loaded_prediction = json.load(file)

print(loaded_prediction)
```
