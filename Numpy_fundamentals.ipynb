Why NumPy is essential

Python lists can store numerical data, but NumPy provides:

Fast numerical computation
Multidimensional arrays
Vectorized operations
Broadcasting
Statistical calculations
Matrix operations
Random-number generation
Foundations for Pandas, scikit-learn, PyTorch, and many ML libraries

1.Python list versus NumPy array

```python
import numpy as np


python_list = [10, 20, 30, 40]

numpy_array = np.array([10, 20, 30, 40])

print("Python list:")
print(python_list)

print("\nNumPy array:")
print(numpy_array)

print("\nTypes:")
print(type(python_list))
print(type(numpy_array))
```

ML Mini Project: Feature Scaling from Scratch

```python
import numpy as np

np.random.seed(42)

X = np.random.randint(
    10,
    100,
    (5,3)
)

print("Original Data")
print(X)

mean = np.mean(X, axis=0)

std = np.std(X, axis=0)

X_scaled = (X - mean) / std

print("\nMean:", mean)


print("\nStd:",std)

print("\nScaled Data:")
print(X_scaled)
```

ML Mini Project: Outlier Detection

```python
import numpy as np

salary = np.array([
    25000,
    28000,
    30000,
    32000,
    35000,
    400000
])

mean = np.mean(salary)
std = np.std(salary)

z_scores = (salary - mean) / std

outliers = salary[np.abs(z_scores) > 2]

print("Outliers:", outliers)
```
