```python
iqr = q3 - q1

print(iqr)
```

10. Outlier Detection

```python
data = np.array([
    10,
    20,
    30,
    40,
    50,
    1000
])

q1 = np.percentile(data,25)
q3 = np.percentile(data,75)

iqr = q3 - q1

lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr

outliers = data[
    (data<lower)|
    (data>upper)
]

print(outliers)
```

12. Normal Distribution

Most important statistical concept in ML.

Characteristics:

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

sns.histplot(
    data,
    kde=True
)

plt.show()
```

13. Z-Score

Used in feature scaling and outlier detection

Rule: |Z| > 3 (Potential Outlier)

Formula: (value - mean)/std

```python
data = np.array([
    10,
    20,
    30,
    40,
    50
])

z_scores = (
    data - np.mean(data)
) / np.std(data)

print(z_scores)
```

ML Mini Project 1: Employee Salary Statistics

```python
import numpy as np

salary = np.array([
    30000,
    35000,
    40000,
    50000,
    60000
])

print("Mean:", np.mean(salary))
print("Median:", np.median(salary))
print("Variance:", np.var(salary))
print("Std:", np.std(salary))
```

ML Mini Project 2: Outlier Detection

```python
import numpy as np

salary = np.array([
    30000,
    35000,
    40000,
    45000,
    50000,
    500000
])

z = (
    salary - np.mean(salary)
)/np.std(salary)

print("Outlier Value:",salary[abs(z)>2])
```
